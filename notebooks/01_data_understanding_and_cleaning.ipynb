```python
# Create cleaning log entries from actual dataset findings

cleaning_records = []

# 1. Record missing-value decisions
missing_counts = df.isnull().sum()

for column, count in missing_counts.items():
    if count > 0:
        cleaning_records.append({
            "issue": "Missing values",
            "affected_column": column,
            "affected_rows": int(count),
            "action_taken": "Retained as missing",
            "reason": "No reliable replacement value is available."
        })

# 2. Duplicate records
cleaning_records.append({
    "issue": "Exact duplicate rows",
    "affected_column": "All columns",
    "affected_rows": int(df.duplicated().sum()),
    "action_taken": "No removal required",
    "reason": "No exact duplicate rows were identified."
})

# 3. Duplicate IDs
cleaning_records.append({
    "issue": "Duplicate order IDs",
    "affected_column": "ID",
    "affected_rows": int(df["ID"].duplicated().sum()),
    "action_taken": "No removal required",
    "reason": "No duplicate order IDs were identified."
})

# 4. Invalid ratings
cleaning_records.append({
    "issue": "Ratings outside assumed 1-5 range",
    "affected_column": "Delivery_person_Ratings",
    "affected_rows": int(invalid_rating_count),
    "action_taken": "Converted invalid ratings to missing values",
    "reason": "The assumed rating scale is 1-5 and the correct ratings cannot be inferred."
})

# 5. Age below 18
cleaning_records.append({
    "issue": "Delivery partner age below 18",
    "affected_column": "Delivery_person_Age",
    "affected_rows": int((df["Delivery_person_Age"] < 18).sum()),
    "action_taken": "Retained pending verification",
    "reason": "The available documentation does not establish a minimum valid age."
})

# 6. Potential outliers
for column in ["Delivery_person_Ratings", "Time_taken (min)"]:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = (
        (df[column] < lower) |
        (df[column] > upper)
    ).sum()

    cleaning_records.append({
        "issue": "Potential IQR outliers",
        "affected_column": column,
        "affected_rows": int(count),
        "action_taken": "Retained",
        "reason": "Statistical outliers are not automatically invalid."
    })

# 7. Zero restaurant coordinates
zero_coordinate_count = (
    (df["Restaurant_latitude"] == 0) &
    (df["Restaurant_longitude"] == 0)
).sum()

cleaning_records.append({
    "issue": "Restaurant coordinates equal to zero",
    "affected_column": "Restaurant_latitude, Restaurant_longitude",
    "affected_rows": int(zero_coordinate_count),
    "action_taken": "Retained pending geographical verification",
    "reason": "The coordinate validity requires further source verification."
})

# 8. Save cleaning log
cleaning_log = pd.DataFrame(cleaning_records)

# Locate project root whether notebook runs from root or notebooks/
project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent

log_path = project_root / "docs" / "cleaning_log.csv"

cleaning_log.to_csv(log_path, index=False)

print("Cleaning log saved successfully!")
print("Location:", log_path.resolve())

display(cleaning_log)
```

```python
import re
import numpy as np
import pandas as pd

def convert_time_to_minutes(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    # Decimal fraction of a 24-hour day
    try:
        numeric_value = float(value)

        if 0 <= numeric_value <= 1:
            return round(numeric_value * 1440)

    except ValueError:
        pass

    # Standard HH:MM or extended HH:MM:SS
    match = re.fullmatch(
        r"(\d{1,2}):(\d{2})(?::(\d{2}))?",
        value
    )

    if match:
        hours = int(match.group(1))
        minutes = int(match.group(2))
        seconds = int(match.group(3) or 0)

        if minutes < 60 and seconds < 60:
            return hours * 60 + minutes + seconds / 60

    return np.nan


# Create new standardized columns
df_clean["Time_Orderd_Minutes"] = (
    df_clean["Time_Orderd"].apply(convert_time_to_minutes)
)

df_clean["Time_Order_picked_Minutes"] = (
    df_clean["Time_Order_picked"].apply(convert_time_to_minutes)
)

print("Time conversion completed!")
```
